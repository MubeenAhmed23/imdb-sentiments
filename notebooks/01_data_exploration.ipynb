# IMDb data exploration

Load and inspect the official train/test splits. This notebook complements the reusable pipeline in `src/`.


```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

sys.path.insert(0, str(Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()))
from src.data import load_imdb

train, test = load_imdb()
print(f'Train: {train.shape}; test: {test.shape}')
```

```python
for name, frame in [('train', train), ('test', test)]:
    print(f'\n{name} label counts:')
    print(frame['label'].value_counts().sort_index().rename(index={0: 'negative', 1: 'positive'}))
    print(f'Duplicate review texts: {frame.text.duplicated().sum():,}')

train['word_count'] = train['text'].str.split().str.len()
train.groupby('label')['word_count'].describe()
```

```python
train.boxplot(column='word_count', by='label', showfliers=False)
plt.title('IMDb review length by sentiment')
plt.suptitle('')
plt.xlabel('Label (0 = negative, 1 = positive)')
plt.ylabel('Words per review')
plt.tight_layout()
plt.show()
```
